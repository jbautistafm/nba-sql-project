# 01 · Exploración inicial del dataset

Primer vistazo a los CSV de `data/raw`: qué archivos hay y cómo es la cabecera de cada tabla.

```python
import pandas as pd
from pathlib import Path

# para ver todas las columnas al hacer head()
pd.set_option("display.max_columns", None)
```

## 1. Qué archivos hay en data/raw
Si sale vacío, primero hay que ejecutar `python src/download_data.py`.

```python
RAW_DIR = Path("../data/raw")

archivos = sorted(RAW_DIR.glob("*.csv"))

for archivo in archivos:
    print(archivo.name)

print("\nTotal:", len(archivos), "archivos")
```

## 2. Cabecera de cada tabla
Para cada CSV saco el número de filas y columnas y las 5 primeras filas.

```python
for archivo in archivos:
    df = pd.read_csv(archivo)
    print("=" * 60)
    print(archivo.name, "->", df.shape[0], "filas,", df.shape[1], "columnas")
    display(df.head())
```
